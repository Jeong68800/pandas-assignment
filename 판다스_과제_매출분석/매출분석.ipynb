### 1. 파일 읽기

- `baemin` 폴더의 파일 31개를 모두 읽어서 하나의 DataFrame으로 만드세요.
- `coupang` 폴더의 파일 31개도 같은 방식으로 읽으세요.
- 파일 목록은 아래 힌트 코드로 가져오고, 반복문으로 하나씩 읽어서 합치세요.

```python
# 배민 파일 하나의 dataframe으로 만들기
baemin_files = sorted(glob.glob("D:/kh261013/eda_repo/pandas_repo/homework/data/baemin/baemin*.csv"))
baemin_list =[]
for file in baemin_files:
    baemin = pd.read_csv(file)
    baemin_list.append(baemin)

baemin_df = pd.concat(baemin_list, ignore_index=True)

baemin_df
```

```python
# 쿠팡 파일 하나의 dataframe으로 만들기
coupang_files = sorted(glob.glob("D:/kh261013/eda_repo/pandas_repo/homework/data/coupang/coupang*.csv"))
coupang_list =[]
for file in coupang_files:
    coupang = pd.read_csv(file)
    coupang_list.append(coupang)

coupang_df = pd.concat(coupang_list, ignore_index=True)

coupang_df
```

```python
# | 내용 | 배민 | 쿠팡 |
# | --- | --- | --- |
# | 주문일시 | `order_at` | `order_time` |
# | 전체금액 | `total_bill` | `total_price` |
# | 광고유입여부 (Y/N) | `ad_yn` | `advertisement_yn` |
# | 광고 관련 | `cpc`: 광고로 들어온 주문 1건당 광고비(원), 광고 아니면 0 | `ad_pct`: 매출 대비 광고비율(%), 숫자만 (5 → 5%), 광고 아니면 0 |
```

### 2. 칼럼명 통일

- 두 플랫폼의 칼럼명을 `주문일시`, `전체금액`, `광고유입여부`로 바꾸세요.
- 어느 플랫폼 데이터인지 알 수 있도록 `플랫폼` 칼럼(`배민` / `쿠팡`)을 추가하세요.
- 두 플랫폼 모두 주문 1건마다 `광고집행비`(원) 칼럼을 만드세요.
    - 배민: `cpc` 그대로
    - 쿠팡: `전체금액 × ad_pct / 100`

```python
baemin_df.rename(columns={
    'order_at' : '주문일시',
    'total_bill' : '전체금액',
    'ad_yn' : '광고유입여부 (Y/N)',        
}, inplace=True)

baemin_df['플랫폼'] = '배민'
baemin_df['광고집행비'] = baemin_df['cpc']
baemin_df
```

```python
coupang_df.rename(columns={
    'order_time' : '주문일시',
    'total_price' : '전체금액',
    'advertisement_yn' : '광고유입여부 (Y/N)',
}, inplace=True)

coupang_df['플랫폼'] = '쿠팡'
coupang_df['광고집행비'] = coupang_df['전체금액'] * coupang_df['ad_pct'] / 100
coupang_df
```

### 3. 데이터 통합

- 두 DataFrame을 하나로 합치세요. (`pd.concat`)
- `주문일시`를 datetime으로 바꾸고, 날짜만 담은 `주문일자` 칼럼을 추가하세요.
- 주문일시 순으로 정렬하세요.

```python
df = pd.concat([baemin_df, coupang_df], ignore_index=True)
df['주문일시'] = pd.to_datetime(df['주문일시'])
# type(df['주문일시'])
df['주문일자'] = df['주문일시'].dt.date
# df['주문일자']
```

### 4. 일자별 매출 집계

- 행은 `주문일자`, 열은 `플랫폼`, 값은 `전체금액` 합계인 표를 만드세요. (`pivot_table`)
- `합계` 열을 추가하세요.

```python
df_table = df.pivot_table(
    index='주문일자',
    columns='플랫폼',
    values='전체금액',
    aggfunc='sum',
)

df_table
```

### 5. 광고 효과 분석

- 플랫폼별로 광고유입 주문이 전체 주문 건수에서 차지하는 **비율(%)**을 구하세요.
- 플랫폼별 **광고집행비** 합계를 구하세요.

```python
ratio = df.groupby('플랫폼')['광고유입여부 (Y/N)'].value_counts(normalize=True).unstack()['Y'] * 100
print(round(ratio, 2))

ad_cost = df.groupby('플랫폼')['광고집행비'].sum()
ad_cost
```

### 6. 식자재비 추출

- 은행 거래내역에서 거래처가 정확히 `식자재마트`인 출금 내역만 골라내세요.
- 일자별 식자재비 합계를 구하세요.

```python
bank_df = pd.read_csv("data/bank_20261001_20261031.csv", encoding="cp949")
print(bank_df)

x = bank_df[bank_df['거래처'] == '식자재마트']
print(x['출금액'])

bank_df['거래일시'] = pd.to_datetime(bank_df['거래일시'])
bank_df['거래일자'] = bank_df['거래일시'].dt.date
y = bank_df.groupby('거래일자')['출금액'].sum()
y
```

### 7. 최종 결과표 저장

- 4번 결과와 6번 결과를 일자 기준으로 합쳐서 아래 칼럼을 가진 표를 만드세요.

```python
# 주문일자. 배민, 쿠팡 합계 식자재비 매출-식자재비

df_table['합계'] = df_table.sum(axis=1)
# df_table

result = df_table.copy()
result['식자재비'] = y

result['매출-식자재비'] = result['합계'] - result['식자재비']
result

```

```python
result.to_csv("result.csv", encoding="utf-8-sig")
```

### 1. inplace=True
* 원본 데이터프레임 자체를 수정해서 덮어씌우는 것

### 2. pct와 cpc 차이
* cpc: 광고로 들어온 주문 1건당 광고비
* ad_pct: 매출 대비 광고비율(%)

### 3. unstack()
* 세로로 길쭉하게 쌓여 있던 결과를 보기 편한 가로/세로 2차원 표 형태로 변환


### 4. value_counts(normalize=True)
* 데이터 개수를 전체 대비 비율(0-1 사이의 값으로 바꿔주는 기능)
