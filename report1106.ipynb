```python
# 1. 나이가 10살 미만인지 검사하는 마스크를 만들어 `head(5)`를 출력하세요.
# 2. 마스크로 10살 미만 승객 수를 출력하세요.

mask = titanic['age'] < 10
print(titanic[mask].head(5))
print("\n")
print(titanic[titanic['age'] < 10].head(5))
```

```python
# 각각 조건을 만족하는 행 수를 출력하세요.

# 1. 3등석 **그리고** 남성 **그리고** 생존한 승객 (`&`)
# 2. 1등석 **또는** 요금 100 이상인 승객 (`|`)
# 3. 혼자 탑승하지 **않은** 승객 (`~`, `alone` 열 사용)

mask01 = (titanic["class"] == 'Third') & (titanic["sex"] == "male") & (titanic["alive"] == "yes")
print(len(titanic[mask01]))

print("\n")

mask02 = (titanic["class"] == 'First') | (titanic["fare"] >= 100)
print(len(titanic[mask02]))

print("\n")

mask03 = ~(titanic["alone"] == True)
print(len(titanic[mask03]))
```

```python
# 1. `isin`으로 Cherbourg 또는 Queenstown에서 탑승한 승객을 골라 `shape`를 출력하세요.
# 2. `between`으로 나이 20~30세(양 끝 포함) 승객을 골라 `shape`와 나이의 min·max를 출력하세요.
# 3. 같은 범위를 양 끝 미포함(`inclusive="neither"`)으로 바꿔 `shape`를 출력하세요.

x = titanic['embark_town'].isin(['Cherbourg', 'Queenstown'])
print(titanic[x].shape)

print("\n")

y = titanic['age'].between(20,30)
print(titanic[y].shape)
print(titanic[y]['age'].min())
print(titanic[y]['age'].max())

print("\n")

z = titanic["age"].between(20, 30, inclusive="neither")
print(titanic[z].shape)



```

```python
# 12살 이하 그리고 3등석 승객의 age, sex, survived 열만 kids 변수에 담고,
#  행 수와 생존률(survived 평균, 소수 둘째 자리)을 출력하세요.

k = (titanic['age'] <= 12) & (titanic['class'] == 'Third')
kids = (titanic.loc[k, ['age', 'sex', 'survived']]) 
print(len(kids))

print("\n")

survived_rate = kids['survived'].mean()
survived_rate = round(survived_rate, 2)
print(survived_rate)
 
```

```python
# 1. 변수 `min_fare = 50`을 만들고, `query`로 Cherbourg 탑승 
# **그리고** 요금이 `min_fare` 이상인 승객의 `shape`를 출력하세요.
# 2. 5-1번 조건을 `query`로 다시 써서 행 수가 같은지 확인하세요.

min_fare = 50
x = titanic.query(("embark_town == 'Cherbourg' and fare > @min_fare"))
print(x.shape)

print("\n")

mask03 = ~(titanic["alone"] == True)
print(len(titanic[mask03]))
y = titanic.query("alone != True")
print(len(y))

```

```python
# 1. `pclass` 오름차순, 같은 등급 안에서는 `fare` 내림차순으로 정렬해 
# `pclass`, `fare`, `sex`, `age` 열의 상위 3행을 출력하세요.
# 2. 요금 내림차순으로 정렬한 결과를 변수에 담고, `iloc[0]`과 `loc[0]`의 `fare` 값을 각각 출력하세요. 
# 두 값이 다른 이유를 주석으로 쓰세요.

x = titanic.sort_values(
    by=['pclass', 'fare'],
    ascending=[True, False]
)
result = x[["pclass", "fare", "sex", "age"]].head(3)
print(result)

print("\n")

y = titanic.sort_values(by='fare', ascending=False)
print(y.iloc[0]['fare'])
print("\n")
print(y.loc[0]['fare'])
# iloc[0]는 정렬된 결과의 0번째 위치의 영역을 가져오고 loc[0]는 원본에서 0번째 위치의 영역을 가져오기 때문(질문해야됨)
```

```python
# 질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

# 1. 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
# 2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
# 3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
# 4. 등급만 3등석으로 바꿔 생존률을 구하시오.

mask01 = ((titanic['sex'] == 'male') & (titanic['age'].between(20,40)) & (titanic['pclass'].isin([1, 2])))
x = titanic.loc[mask01, ['pclass', 'age', 'fare', 'survived']]
print(len(x))

print("\n")

survived_rate01 = x["survived"].mean()
survived_rate01 = round(survived_rate01, 2)
print(survived_rate01)

print("\n")

print(x.sort_values(by='fare', ascending=False).head(3))

print("\n")

y = titanic.query("sex == 'male' and 20<= age <= 40 and pclass in [1, 2]")
print(len(y))

print("\n")

mask02 = ((titanic['sex'] == 'male') & (titanic['age'].between(20,40)) & (titanic['pclass'].isin([3])))
z = titanic.loc[mask02, ['pclass', 'age', 'fare', 'survived']]
survived_rate02 = z["survived"].mean()
survived_rate02 = round(survived_rate02, 2)
print(survived_rate02)
```
